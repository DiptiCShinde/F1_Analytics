### Creating catalog

```python
%sql
create catalog f1_catalog
```

```python
%sql
USE CATALOG f1_catalog;

CREATE SCHEMA IF NOT EXISTS bronze
MANAGED LOCATION 'abfss://bronze@storagef12.dfs.core.windows.net/';

CREATE SCHEMA IF NOT EXISTS silver
MANAGED LOCATION 'abfss://silver@storagef12.dfs.core.windows.net/';

CREATE SCHEMA IF NOT EXISTS gold
MANAGED LOCATION 'abfss://gold@storagef12.dfs.core.windows.net/';
```
